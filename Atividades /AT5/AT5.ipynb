```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregando o dataset diretamente da URL oficial para facilitar
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Visualizando as primeiras linhas
df.head()
```

```python
# Quantidade de linhas e colunas
print(f"O dataset possui {df.shape[0]} linhas e {df.shape[1]} colunas.")

# Tipo de dado de cada variável
print("\nTipos de dados de cada variável:")
print(df.dtypes)
```

```python
# Gera estatísticas como média, desvio padrão, min e max
df.describe()
```

```python
# Tabela de frequência
print("Frequência de Sobrevivência (0 = Não, 1 = Sim):")
print(df['Survived'].value_counts())

# Gráfico
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Survived', palette='viridis')
plt.title('Distribuição da Variável Survival')
plt.show()

# Explicação: O gráfico de barras (countplot) foi escolhido por ser ideal para
# comparar categorias discretas, permitindo visualizar rapidamente a diferença
# entre o número de sobreviventes e não sobreviventes.
```

```python
total_sobreviventes = df['Survived'].sum()
classe_frequente = df['Pclass'].mode()[0]

print(f"Quantidade de pessoas que sobreviveram: {total_sobreviventes}")
print(f"A classe mais frequente entre os passageiros é a: {classe_frequente}ª classe")
```

```python
# Tabela de frequência cruzada
tabela_sexo_survival = pd.crosstab(df['Sex'], df['Survived'])
print("Tabela de Frequência (Sexo vs Sobrevivência):")
print(tabela_sexo_survival)

# Gráfico
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Sex', hue='Survived', palette='magma')
plt.title('Relação entre Sexo e Sobrevivência')
plt.show()

# Resposta: Sim, as mulheres sobreviveram mais que os homens.
```

```python
# Gráfico de distribuição (Histograma com curva)
plt.figure(figsize=(8,5))
sns.histplot(df['Age'].dropna(), kde=True, color='skyblue')
plt.title('Distribuição das Idades dos Passageiros')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()

# Respostas sobre o gráfico:
# 1. A distribuição não é perfeitamente simétrica; ela possui uma leve "cauda" para a direita (assimetria positiva).
# 2. A maioria dos passageiros era JOVEM (concentrada entre 20 e 40 anos), com um número menor de idosos.
```
